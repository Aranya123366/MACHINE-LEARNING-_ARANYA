```python
from sklearn.datasets import load_diabetes
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
X,y=load_diabetes(return_X_y=True)
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=2)
print(X_train.shape)
print(X_test.shape)
```

```python
class GDRegressor:
  def __init__(self,learning_rate=0.01,epoches=1000):
    self.coef_=None
    self.intercept_=None
    self.lr=learning_rate
    self.epoches=epoches

  def fit(self,X_train,y_train):
     #init your coef
     self.intercept_=0
     self.coef_=np.ones(X_train.shape[1])

     for i in range(self.epoches):
        y_hat=np.dot(X_train,self.coef_)+self.intercept_
        intercept_der=(-2/y_train.shape[0])*np.sum(y_train-y_hat)
        self.intercept_=self.intercept_-self.lr*intercept_der

        coef_der=(-2/X_train.shape[0])*np.dot((y_train-y_hat),X_train)
        self.coef_=self.coef_-self.lr*coef_der
     print(self.coef_,self.intercept_)
  def predict(self,X_test):
    return self.intercept_+np.dot(self.coef_,X_test)
```

```python
gdr=GDRegressor()
```

```python
gdr.fit(X_train,y_train)
```

```python
class SGDRegressor:
  def __init__(self,learning_rate=0.01,epoches=300):
    self.intercept_=None
    self.coef_=None
    self.lr=learning_rate
    self.epoches=epoches
  def fit(self,X_train,y_train):
     #init your coef
     self.intercept_=0
     self.coef_=np.ones(X_train.shape[1])

     for i in range(self.epoches):
        for j in range(X_train.shape[0]):
          idx=np.random.randint(0,X_train.shape[0])
          y_hat=np.dot(X_train[idx],self.coef_)+self.intercept_
          intercept_der=(-2)*(y_train[idx]-y_hat)
          self.intercept_=self.intercept_-self.lr*intercept_der

          coef_der=(-2)*np.dot((y_train[idx]-y_hat),X_train[idx])
          self.coef_=self.coef_-self.lr*coef_der

     print(self.coef_,self.intercept_)
  def predict(self,X_test):
    return self.intercept_+np.dot(self.coef_,X_test)
```

```python
sgd=SGDRegressor()
sgd.fit(X_train,y_train)
```

```python
class MBGDRegressor:
  def __init__(self,batch_size,learning_rate=0.01,epoches=1000):
    self.coef_=None
    self.intercept_=None
    self.lr=learning_rate
    self.epoches=epoches
    self.batch_size=batch_size

  def fit(self,X_train,y_train):
     #init your coef
     self.intercept_=0
     self.coef_=np.ones(X_train.shape[1])

     for i in range(self.epoches):
      for j in range(int(X_train.shape[0]/self.batch_size)):
        idx=random.sample(range(X_train.shape[0]),self.batch_size)
        y_hat=np.dot(X_train[idx],self.coef_)+self.intercept_
        intercept_der=(-2/y_train.shape[0])*np.sum(y_train[idx]-y_hat)
        self.intercept_=self.intercept_-self.lr*intercept_der

        coef_der=(-2/X_train.shape[0])*np.dot((y_train[idx]-y_hat),X_train[idx])
        self.coef_=self.coef_-self.lr*coef_der
     print(self.coef_,self.intercept_)
  def predict(self,X_test):
    return self.intercept_+np.dot(self.coef_,X_test)
```

```python
mbr=MBGDRegressor(batch_size=int(X_train.shape[0]/10))
mbr.fit(X_train,y_train)
```
